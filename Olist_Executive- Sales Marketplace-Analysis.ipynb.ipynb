Outlier Detection

```python
# 2. BOXPLOT: Visually check for price and freight outliers
# Run this in your script or notebook to see the chart immediately
df3.boxplot(column=['price', 'freight_value'])
plt.title("Checking for E-commerce Outliers")
plt.show()
```

```python
# 3. IQR METHOD: Flag extreme prices
q1 = df3['price'].quantile(0.25)
q3 = df3['price'].quantile(0.75)
iqr = q3 - q1

lower_limit = q1 - (1.5 * iqr)
upper_limit = q3 + (1.5 * iqr)
```

```python
# Filter rows where price is an IQR outlier
price_iqr_outliers = df3[(df3['price'] < lower_limit) | (df3['price'] > upper_limit)]
```

```python
# 4. Z-SCORE METHOD: Flag extreme freight values
# Calculates how many standard deviations away each shipping cost is
df3['freight_z_score'] = stats.zscore(df3['freight_value'])
```

```python
# Standard threshold is 3, but for small datasets, 2 works best
freight_z_outliers = df3[df3['freight_z_score'].abs() > 2]
```

```python
# 5. PRINT RESULTS
print("--- Price Outliers (Using IQR) ---")
print(price_iqr_outliers[['order_id', 'price']])

print("\n--- Freight Outliers (Using Z-Score) ---")
print(freight_z_outliers[['order_id', 'freight_value', 'freight_z_score']])
```

Feature Engineering

```python
# PHASE 1: FINANCIAL FEATURES
#Revenue: Total cash coming in for the items
df3['Revenue'] = df3['price'] * df3['order_item_id']
```

```python
# Profit Margin: Profit percentage per order
df3['Profit'] = df3['Revenue'] - df3['freight_value']
df3['Profit_Margin'] = (df3['Profit'] / df3['Revenue']) * 100
```

```python
# PHASE 2: TIME & CALENDAR FEATURES
# Shipping Days: How long the fulfillment window is

# Merge df6 into df3 to get order_purchase_timestamp
df3 = pd.merge(df3, df6[['order_id', 'order_purchase_timestamp']], on='order_id', how='left')

# Ensure both columns are datetime objects
df3['shipping_limit_date'] = pd.to_datetime(df3['shipping_limit_date'])
df3['order_purchase_timestamp'] = pd.to_datetime(df3['order_purchase_timestamp'])

# Calculate Shipping_Days
df3['Shipping_Days'] = (df3['shipping_limit_date'] - df3['order_purchase_timestamp']).dt.days
```

```python
# Date breakdowns
df3['Month'] = df3['order_purchase_timestamp'].dt.month
df3['Quarter'] = df3['order_purchase_timestamp'].dt.quarter
df3['Year'] = df3['order_purchase_timestamp'].dt.year
df3['Week'] = df3['order_purchase_timestamp'].dt.isocalendar().week
```

```python
# Weekend: Binary flag (1 if Saturday/Sunday, 0 if weekday)
df3['Weekend'] = df3['order_purchase_timestamp'].dt.dayofweek.apply(lambda x: 1 if x >= 5 else 0)
```

```python
# Order Size: Categorizing order size based on items bought
df3['Order_Size'] = pd.cut(df3['order_item_id'], bins=[0, 1, 3, float('inf')], labels=['Small', 'Medium', 'Large'])
```

```python
# PHASE 4: ADVANCED CUSTOMER METRICS
# Merge customer_id into df3 from df6
df3 = pd.merge(df3, df6[['order_id', 'customer_id']], on='order_id', how='left')

# Calculate total historic spend and total orders per customer
customer_stats = df3.groupby('customer_id').agg(
    total_spend=('Revenue', 'sum'),
    total_orders=('order_id', 'count')
).reset_index()
```

```python
# High Value Customer: Spent more than the top 75% of users (e.g., threshold set to $100 here)
customer_stats['High_Value_Customer'] = customer_stats['total_spend'].apply(lambda x: 1 if x >= 100 else 0)
```

```python
# Repeat Customer: Has placed more than 1 distinct order
customer_stats['Repeat_Customer'] = customer_stats['total_orders'].apply(lambda x: 1 if x > 1 else 0)
```

```python
# Merge behavioral flags back into main DataFrame
df3 = df3.merge(customer_stats[['customer_id', 'High_Value_Customer', 'Repeat_Customer']], on='customer_id', how='left')
```

```python
print(df3[['order_id', 'Revenue', 'Profit_Margin', 'Shipping_Days', 'Weekend', 'High_Value_Customer']])
```

```python
# 1. CUSTOMER SUMMARY TABLE
# =====================================================================
# Groups data by each unique customer to see their buying habits.
customer_summary = df3.groupby('customer_id').agg(
    total_orders=('order_id', 'nunique'),
    total_items_bought=('order_item_id', 'sum'),
    total_spend=('Revenue', 'sum'),
    total_profit=('Profit', 'sum'),
    average_order_value=('Revenue', 'mean')
).reset_index()
```

```python
# 2. PRODUCT SUMMARY TABLE
# =====================================================================
# Shows which specific products are generating the most cash and volume.
product_summary = df3.groupby('product_id').agg(
    units_sold=('order_item_id', 'sum'),
    gross_revenue=('Revenue', 'sum'),
    total_profit=('Profit', 'sum')
).reset_index()
```

```python
# 3. MONTHLY SALES TABLE
# =====================================================================
# Tracks sales performance over time (perfect for line charts).
monthly_sales = df3.groupby(['Year', 'Month']).agg(
    total_revenue=('Revenue', 'sum'),
    total_orders=('order_id', 'nunique'),
    avg_shipping_time=('Shipping_Days', 'mean')
).reset_index().sort_values(by=['Year', 'Month'])
```

```python
# 4. REGIONAL SALES TABLE
# =====================================================================
# Breaks down performance by location (assuming you have a 'seller_region' or 'customer_region')
regional_sales = df3.groupby('seller_id').agg(
    total_sales=('Revenue', 'sum'),
    total_orders=('order_id', 'nunique'),
    avg_freight_cost=('freight_value', 'mean')
).reset_index()
```

```python
# 5. RETURNS SUMMARY TABLE
# =====================================================================
# Assuming you have a 'status' column where 'canceled' or 'returned' is tracked.
# If you don't have a return column, this creates a mock column first to demonstrate:
if 'order_status' not in df3.columns:
    df3['order_status'] = np.random.choice(['delivered', 'returned'], size=len(df3), p=[0.9, 0.1])

returns_summary = df3[df3['order_status'] == 'returned'].groupby('product_id').agg(
    total_returned_units=('order_item_id', 'sum'),
    lost_revenue=('Revenue', 'sum')
).reset_index()
```

```python
# 6. CUSTOMER LIFETIME VALUE (CLV) TABLE
# =====================================================================
# A standard business calculation: Lifespan * Average Purchase Value
customer_clv = df3.groupby('customer_id').agg(
    total_revenue=('Revenue', 'sum'),
    purchase_frequency=('order_id', 'nunique'),
    customer_age_months=('Shipping_Days', lambda x: (x.max() - x.min()) / 30 if x.max() != x.min() else 1)
).reset_index()

# Calculate basic CLV metric
customer_clv['CLV'] = customer_clv['total_revenue'] * customer_clv['purchase_frequency']
```

```python
# 7. SALES BY CATEGORY TABLE
# =====================================================================
# Summarizes performance by high-level departments (assuming you have a category column)
if 'product_category' not in df3.columns:
    df3['product_category'] = np.random.choice(['Electronics', 'Apparel', 'Home', 'Beauty'], size=len(df3))

sales_by_category = df3.groupby('product_category').agg(
    total_revenue=('Revenue', 'sum'),
    total_profit=('Profit', 'sum'),
    average_item_price=('price', 'mean')
).reset_index()
```
